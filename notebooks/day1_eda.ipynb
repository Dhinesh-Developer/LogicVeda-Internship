# renaming the columns


```python
# Remove extras columns
df = df.dropna(axis=1)
```

```python
# Adding the Proper Column Names

# unit_id => which machine
# cycle => time step
# sensor_* => machine readings
# Each machine runs until failure
cols = ['unit_id','cycle']+\
        [f'op_setting_{i}' for i in range(1,4)] + \
        [f'sensor_{i}' for i in range(1,22)]

df.columns = cols
```

```python
df.head()
```

```python
machine_1 = df[df['unit_id'] == 1]
```

```python
plt.plot(machine_1['cycle'], machine_1['sensor_2'])
plt.title('Machine 1 sensor Behaviour')
plt.xlabel('time (cycle)')
plt.ylabel('Sensor value')
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(machine_1['cycle'], machine_1['sensor_2'], label='Sensor 2 (Increasing)')
plt.plot(machine_1['cycle'], machine_1['sensor_7'], label='Sensor 7 (Decreasing)')
plt.plot(machine_1['cycle'], machine_1['sensor_10'], label='Sensor 10 [Neutral]')

plt.legend()
plt.title("Sensor Behaviour Comparsion")
plt.xlabel("Cycle")
plt.ylabel("Value")
plt.show()
```

```python
plt.plot(machine_1['cycle'], machine_1['sensor_10'])
plt.title("Sensor 10 Behaviour")
plt.show()
```

# Observation 
- Sensor 2 shows an increasing trend indicating possible degradation
- Sensor 7 shows a decreasing trend suggesting reduced efficiency
- Sensor 10 remains stable, indicating not all sensors are sensitive to failure
- Different sensors behave differently, showing multi-dimensional machine degradtion
- Overall data follows a time based degradtion pattern

- These patterns indicate that sensor behaviour can be used to estimate machine health and predict 'Remaining Useful Life(RUL)'

```python
# RUL finding
df['RUL'] = df.groupby('unit_id')['cycle'].transform('max') - df['cycle']
```

```python
machine_1 = df[df['unit_id'] == 1]

plt.plot(machine_1['cycle'], machine_1['RUL'])
plt.title('Remaining Useful Life')
plt.xlabel('cycle')
plt.ylabel('RUL')
plt.show()
```
