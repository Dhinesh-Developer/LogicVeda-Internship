```python
from kafka import KafkaProducer
import json
import time
import random
from datetime import datetime
```

```python
producer = KafkaProducer(
    bootstrap_servers = 'localhost:9092',
    value_serializer= lambda v: json.dumps(v).encode('utf-8')
)
```

```python
while True:
    data = {
        "machine_id": 1,
        "sensor_1": random.uniform(20, 80),
        "sensor_2": random.uniform(10, 100),
        "sensor_3": random.uniform(30, 90),
        "timestamp": time.time()
    }

    producer.send("sensor-data", value=data)
    print("Sent:", data)

    time.sleep(2)
```
