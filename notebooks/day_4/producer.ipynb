```python
from kafka import KafkaProducer
from faker import Faker
import json
import time
```

```python
producer = KafkaProducer(
    bootstrap_servers = 'localhost:9092',
    value_serializer = lambda v:json.dumps(v).encode('utf-8')
)

fake = Faker()
```

```python
while True:
    data = {
        "name":fake.name(),
        "email":fake.email(),
        "city":fake.city(),
        "timestamp":fake.time()
    }

    print("Sending:",data)
    producer.send('user-data', value=data)
    time.sleep(2)
```
