```python
# postgres connection
con = psycopg2.connect(
    dbname = "kafka_db",
    user = 'postgres',
    password = 'arise',
    host = 'localhost',
    port = '5432'
)

cursor = con.cursor()
```

```python
consumer = KafkaConsumer(
    'user-data',
    bootstrap_servers='localhost:9092',
    auto_offset_reset='earliest',
    value_deserializer = lambda x:json.loads(x.decode('utf-8'))
)

print("Listening and storing data...")
```

```python
for message in consumer:
    data = message.value

    print("Saving:", data)

    try:
        cursor.execute(
        "INSERT INTO users (name, email, city) VALUES (%s, %s, %s)",
        (data['name'], data['email'], data['city'])
        )
        con.commit()

    except Exception as e:
        print("Error:", e)
        con.rollback()  
```
